```python
import pandas as pd

df = pd.read_csv('predictedk.csv')

porc_por_col = df.apply(lambda col: col.value_counts(normalize=True))
porc_por_col = porc_por_col.fillna(0)
promedio_porcentaje = porc_por_col.mean(axis=1) * 100

print(promedio_porcentaje)
```
