```python
import pandas as pd

dataset = pd.read_csv('3203_merged_DrugBank_ZINC_chEMBL_KEGG.csv')

atc = dataset['ATC Codes']
```

```python
# Converts a string that simulates a list to a real list
def convert_string_list(element):
    # Delete [] of the string
    element = element[0:len(element)]
    # Create a list that contains each code as e.g. 'A'
    ATC_list = list(element.split('; '))
    for index, code in enumerate(ATC_list):
        # Delete '' of the code
        ATC_list[index] = code[0:len(code)]
    return ATC_list
```

```python
info_atc = pd.DataFrame(columns = ['Number of ATC Codes'])
for codes in atc:
    list_codes = convert_string_list(codes)
    number = len(list_codes)
    info_atc.loc[len(info_atc)] = number

info_atc
```

```python
info_atc.to_csv('info_atc.csv')
```

```python
from collections import Counter

counter_atc = Counter(info_atc['Number of ATC Codes'])
counter_atc
```

```python
split_dataset = pd.read_csv('splitATC.csv')

level1 = split_dataset['ATC_level1']

level2 = split_dataset['ATC_level2']

level3 = split_dataset['ATC_level3']

level4 = split_dataset['ATC_level4']
```

```python
# Converts a string that simulates a list to a real list
def convert_string_list2(element):
    # Delete [] of the string
    element = element[1:len(element)-1]
    # Create a list that contains each code as e.g. 'A'
    ATC_list = list(element.split(', '))
    for index, code in enumerate(ATC_list):
        # Delete '' of the code
        ATC_list[index] = code[1:len(code)-1]
    return ATC_list
```

```python
counter_atc1 = Counter(
    code 
    for codes in level1
    for code in convert_string_list2(codes)
)
print(dict(counter_atc1))
print(len(counter_atc1))
suma = 0
for cant in counter_atc1.values():
    suma += cant
print(suma)
```
